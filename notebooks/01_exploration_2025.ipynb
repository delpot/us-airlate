# 00 - First look at the 2025 data

**Goal:** get a first understanding of the 2025 dataset by looking at a single monthly file (January), before working with the whole year.

## 1. Setup

```python
from pathlib import Path
import pandas as pd

pd.set_option("display.max_columns", None)

DATA_DIR = Path("../data/raw/flight_delays_2025")
MONTH_FILE = DATA_DIR / "flight_delays_2025_01.csv"
```

## 2. Available files

```python
files = sorted(DATA_DIR.glob("*.csv"))

file_sizes = pd.DataFrame({
    "file": [f.name for f in files],
    "size_mb": [round(f.stat().st_size / 1024**2, 1) for f in files],
})

print(f"{len(files)} files, {file_sizes['size_mb'].sum():.0f} MB in total")
file_sizes
```

1 file = 1 month

## 3. Explore first month

```python
df = pd.read_csv(MONTH_FILE, low_memory=False)
```
